Plot only the weapon sales per day of the week.

Customize the graph to add a title, and x and y labels.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
day = np.array(['Mon', 'Tue', 'Wed', 'Thur', 'Fri', 'Sat', 'Sun'])
weapon = np.array([5,4,6,3,7,9,6])

df = pd.DataFrame({'Day':day, 'Weapon':weapon})

plt.figure()
plt.plot(df['Day'], df['Weapon'], label='Weapon sales', linestyle='--', color='blue')
plt.legend()
plt.title('Weapon sales per day of the week')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.show()
```
