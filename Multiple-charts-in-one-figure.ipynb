```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Data for 4 students in a classroom in 4 subjects.

```python
df = pd.DataFrame({
    "Exam": ["CS", "Physics", "Calculus", "Anatomy"],
    "Alice": [90, 85, 80, 88],
    "Bob": [85, 87, 82, 90],
    "Charlie": [88, 90, 84, 86],
    "Diana": [92, 88, 85, 89]
})
```

Charts:

```python
fig, axes = plt.subplots(2, 2, figsize=(10,10), sharex=True, sharey=True) # 2 rows, 2 columns, share x axe and y axe
plt.suptitle('Students Scores') # Main title
plt.xlabel=('Subjects')
plt.ylabel=('Score')

axes[0][0].bar(df['Exam'], df['Alice'], color = 'skyblue')
axes[0][0].set_title('Alice´s Scores')

axes[0][1].bar(df['Exam'], df['Bob'], color = 'navy')
axes[0][1].set_title('Bob´s Scores')

axes[1][0].bar(df['Exam'], df['Charlie'], color = 'lime')
axes[1][0].set_title('Charlie´s Scores')

axes[1][1].bar(df['Exam'], df['Diana'], color = 'pink')
axes[1][1].set_title('Diana´s Scores')

plt.show()
```
