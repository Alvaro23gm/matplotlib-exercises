```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
toppings = np.array(['Pepperoni', 'Cheese', 'Mushroom', 'Pineapple', 'Prosciutto', 'Artichoke'])
orders = np.array([40, 25, 20, 15, 4, 5])
```

Creating a pie chart:

```python
plt.pie(orders, labels=toppings)
plt.title('Favorite Pizza Toppings')
plt.show()
```

We can add percentages to the chart:
- %d%%: no decimals
- %.1f%%: 1 decimal
- %.2f%%: 2 decimals

```python
plt.pie(orders, labels=toppings, autopct='%d%%')
plt.title('Favorite Pizza Toppings')
plt.show()
```

We can explode a slice:

```python
explode = [0.1, 0, 0, 0, 0, 0] # explode the first slice by 10% of radius
plt.pie(orders, labels=toppings, autopct='%d%%', explode=explode)
plt.title('Favorite Pizza Toppings')
plt.show()
```
