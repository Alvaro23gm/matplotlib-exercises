```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({'month': ['Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],'adoptions': [45000, 52000, 48000, 43000, 38000, 41000]})

plt.figure()
plt.plot(df['month'], df['adoptions'], label='cat adoptions', color='yellow',linestyle='-.')
plt.title('Cat adoptions in the U.S.')
plt.xlabel('Months')
plt.ylabel('Amount of adoptions')
plt.legend()
plt.show()
```
