```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Information:

```python
months = np.array(['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September','October', 'November', 'December'])
electric_amount = np.array([145.2,128.5,95.4,88.1,110.3,115.6,102.4,89.2,91,108.7,134.9,10])
gas_amount = np.array([22.1,24,35.8,68.4,112.5,125.8,98.1,65.3,42,28.4,23.1,21.5])
total_amount = electric_amount + gas_amount
```

Graphic:

```python
plt.figure(figsize=(15, 10))
plt.plot(months, electric_amount, label='Electric bills')
plt.plot(months, gas_amount, label='Gas bills')
plt.plot(months, total_amount, label='Total amount')
plt.legend()
plt.show()
```

DataFrame and analysis:

```python
data = {'Month': months, 'Electric Amount': electric_amount, 'Gas Amount': gas_amount, 'Total Amount': total_amount}
df_data = pd.DataFrame(data)

max_electric = df_data['Electric Amount'].max()
min_electric = df_data['Electric Amount'].min()
max_gas = df_data['Gas Amount'].max()
min_gas = df_data['Gas Amount'].min()
total_electric = df_data['Electric Amount'].sum()
total_gas = df_data['Gas Amount'].sum()
total_year_amount = total_electric + total_gas

print(f'Month with the highest electric bills: ${max_electric} USD')
print(f'Month with the lowest electric bills: ${min_electric} USD')
print(f'Month with the highest gas bills: ${max_gas} USD')
print(f'Month with the lowest gas bills: ${min_gas} USD')
print(f'Total amount of the year: ${total_year_amount} USD')

df_data
```
