```python
import numpy as np
import matplotlib.pyplot as plt
```

Values:

```python
games = np.array([1,2,3,4,5]).astype(int)

bulls_points = np.array([89,104,132,203,115])
lakers_points = np.array([102,110,95,108,115])
celtics_points = np.array([98,105,100,112,107])
warriors_points = np.array([110,99,105,120,115])
```

Graphic:

```python
plt.figure()
plt.plot(games, bulls_points, label='Bulls',color='red',linestyle='-')
plt.plot(games, lakers_points, label='Lakers',color='yellow',linestyle='-.')
plt.plot(games, celtics_points, label='Celtics',color='green',linestyle=':')
plt.plot(games, warriors_points, label='Warriors',color='blue',linestyle='--')
plt.legend()
plt.xlabel('Games')
plt.ylabel('Points')
plt.title('NBA Teams')
plt.legend()
plt.show()
```
